# Milestone 3

## Creating Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# defining 2 cols
# numerical col
num_cols = ['age','resting_blood_pressure','serum_cholestoral','maximum_heart_rate_achieved',
            'oldpeak','number_of_major_vessels']

# categorical col
cat_col = ['sex','chest','thal','slope']
# we didn't include remaining 3 cols in cat_col as they already have 0/1: so we let them passthrough into columntransformer
```

```python
# creating pipeline

#Numerical pipeline
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()) # scaling num values as logistic regression +SVMs are sensitive to feature magnitude
])


#Categorical Pipeline
cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehotencoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine pipelines above
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_col)
], remainder='passthrough')
```

```python
# fitting and transforming data
X_train_prepared = preprocessor.fit_transform(X_train)

```

```python
# verifying if pipeline worked
X_train_prepared.shape
```

Findings for Milestone 3:
1. Pipeline was built that imputes any missing values (there aren't any for this dataset) and scales 6 numerical cols, splits categorical columns into 13 cols in the form of 0/1 using hotencoder.
2. Scaling was important as some models like logistic regression and SVMs are sensitive to feature's magnitude and category columns having values (1,2,3,4 etc.) should be encoded
3. X_train went from (216,13) to (216,21) proved pipeline work successfully.

# Milestone 4

## Training a baseline model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
# for this classification project, we are going to use
# LogisticRegression model and wrap with preprocessor using pipeline

log_reg = Pipeline([
    ('prep', preprocessor),
    ('model', LogisticRegression(max_iter=1000))
])
```

```python
y_train_pred = cross_val_predict(log_reg, X_train, y_train, cv=5)
```

```python
y_train_pred
```

```python
# printing metric functions and checking scores

print("Confusion Matrix:\n", confusion_matrix(y_train, y_train_pred))
print("Precision Score:\n", precision_score(y_train, y_train_pred))
print("Recall Score:\n", recall_score(y_train, y_train_pred))
print("F1 Score:\n", f1_score(y_train, y_train_pred))
```

Findings for milestone 4:

 Trained baseline model Logistic Regression (CV fold=5) achieved 83.7% precision, 80.2% recall and 81.9% f1 score, correctly catching 77 out of 96 patients while missing 19 patients.
