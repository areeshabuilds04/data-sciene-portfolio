Signal De-noising

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#creating array of time values: 0 - 10
time = np.linspace(0,10,100)
```

```python
#generate a sin wave
signal = np.sin(time)
signal
```

```python
#create noise
import random
noise1 = np.random.rand(len(time))
noise = noise1 + signal
```

```python
#plot the noisy wave
plt.plot(signal, label='Signal')
plt.plot(noise, label = 'Noise')
plt.legend()

plt.xlabel('Time')
plt.ylabel('Signal')
plt.show()
```

```python
#de-noising using pandas
df_noise = pd.DataFrame(noise)
df_noise
```

```python
df_noise.rename(columns={0: 'noise_value'}, inplace=True)
display(df_noise)
```

```python
#smoothing out noise
smooth_noise = df_noise['noise_value'].rolling(window=20).mean()
smooth_noise
```

```python
plt.plot(signal,label='Signal')
plt.plot(noise,label='Noise')
plt.plot(smooth_noise, label = 'Smooth signal')
plt.legend()
plt.show()
```
